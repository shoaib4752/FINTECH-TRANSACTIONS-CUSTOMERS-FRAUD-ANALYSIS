```python
# customer analysis
transactions.merge(customers, on="Customer_ID").groupby("Customer_Segment")["Transaction_Amount"].sum()
```

```python
# acoount type wise transaction value
transactions.merge(customers, on="Customer_ID").groupby("Account_Type")["Transaction_Amount"].sum()
```

```python
# state wise transaction value

transactions.merge(customers, on="Customer_ID").groupby("State")["Transaction_Amount"].sum()
```

```python
# merchant category wise transaction value

transactions.merge(merchants, on="Merchant_ID").groupby("Merchant_Category")["Transaction_Amount"].sum()
```

```python
# risk level wise transaction count

transactions.merge(riskfraud, on="Transaction_ID").groupby("Risk_Level")["Transaction_ID"].count()
```
